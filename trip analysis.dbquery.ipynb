```sql
SELECT 
    CAST(tpep_pickup_datetime AS DATE) AS pickup_date,
    COUNT(VendorID) AS total_trips,
    SUM(passenger_count) AS total_passengers_carried,
    ROUND(SUM(trip_distance), 2) AS total_distance_miles,
    ROUND(SUM(fare_amount), 2) AS total_fare_revenue_usd,
    ROUND(SUM(tip_amount), 2) AS total_tips_collected_usd,
    ROUND(SUM(total_amount), 2) AS total_gross_revenue_usd,
    ROUND(AVG(trip_duration_minutes), 2) AS avg_trip_duration_minutes,
    -- Safe division for tip percentage
    ROUND(
        COALESCE(SUM(tip_amount) / NULLIF(SUM(fare_amount), 0) * 100, 0), 2
    ) AS avg_tip_percentage
FROM dev_nyc.silver.taxi_duartion -- Replace with your actual silver table name
GROUP BY CAST(tpep_pickup_datetime AS DATE)
ORDER BY pickup_date DESC;



```
