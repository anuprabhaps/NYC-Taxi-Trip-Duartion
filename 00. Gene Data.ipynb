## Generating New York Taxi trip data

```python
import csv
import os
import random
from datetime import datetime, timedelta

def nyc_taxi_stream_generator(num_records=10000):
    start_date = datetime(2025, 1, 1)
    
    for i in range(num_records):
        pickup_time = start_date + timedelta(
            days=random.randint(0, 28),
            hours=random.randint(0, 23),
            minutes=random.randint(0, 59)
        )
        dropoff_time = pickup_time + timedelta(minutes=random.randint(5, 45))
        
        fare = round(random.uniform(5.0, 60.0), 2)
        tip = round(random.choice([0, fare * 0.15, fare * 0.20]), 2)
        mta_tax = 0.50
        improvement_surcharge = 0.30
        total = round(fare + tip + mta_tax + improvement_surcharge, 2)

        yield {
            "VendorID": random.choice([1, 2]), # <-- Fixed missing parameters
            "tpep_pickup_datetime": pickup_time.strftime("%Y-%m-%d %H:%M:%S"),
            "tpep_dropoff_datetime": dropoff_time.strftime("%Y-%m-%d %H:%M:%S"),
            "passenger_count": random.randint(1, 6),
            "trip_distance": round(random.uniform(0.5, 18.5), 2),
            "RatecodeID": 1,
            "store_and_fwd_flag": random.choice(["N", "Y"]),
            "PULocationID": random.randint(1, 263),
            "DOLocationID": random.randint(1, 263),
            "payment_type": random.choice([1, 2, 3]), # <-- Fixed missing parameters
            "fare_amount": fare,
            "extra": random.choice([0.0, 0.5, 1.0]), # <-- Fixed missing parameters
            "mta_tax": mta_tax,
            "tip_amount": tip,
            "tolls_amount": 0.0,
            "improvement_surcharge": improvement_surcharge,
            "total_amount": total
        }

# Set the path matching your specific volume layout
volume_path = "/Volumes/dev_nyc/landing/nyc_taxi_trip_duration/"
os.makedirs(volume_path, exist_ok=True)
csv_file_path = os.path.join(volume_path, "generated_trips_raw.csv")

headers = [
    "VendorID", "tpep_pickup_datetime", "tpep_dropoff_datetime", "passenger_count",
    "trip_distance", "RatecodeID", "store_and_fwd_flag", "PULocationID", "DOLocationID",
    "payment_type", "fare_amount", "extra", "mta_tax", "tip_amount", "tolls_amount",
    "improvement_surcharge", "total_amount"
]

print(f"Writing corrected file to: {csv_file_path}")
with open(csv_file_path, mode="w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=headers)
    writer.writeheader()
    for record in nyc_taxi_stream_generator(num_records=25000):
        writer.writerow(record)

print("File regeneration successfully complete!")
```
