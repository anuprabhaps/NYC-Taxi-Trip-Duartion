```sql
SELECT 
    CONCAT('Zone ', PULocationID, ' ➡️ Zone ', DOLocationID) AS transit_corridor,
    VendorID,
    SUM(trip_count) AS trip_count,
    ROUND(SUM(total_revenue), 2) AS corridor_revenue_usd,
    ROUND(SUM(total_distance_miles) / NULLIF(SUM(total_duration_minutes) / 60.0, 0), 2) AS corridor_avg_speed_mph
FROM 
    dev_nyc.gold.fact_vendor_performance
GROUP BY 
    PULocationID, DOLocationID, VendorID
ORDER BY 
    corridor_revenue_usd DESC
LIMIT 10;
```
